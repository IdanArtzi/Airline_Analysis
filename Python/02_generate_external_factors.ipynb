```python
fuel = calendar.copy()

fuel["Fuel_Index"] = 100
```

```python
fuel.loc[
    (fuel["Date"] >= "2020-03-01") &
    (fuel["Date"] <= "2020-12-31"),
    "Fuel_Index"
] = 75

fuel.loc[
    (fuel["Date"] >= "2021-01-01") &
    (fuel["Date"] <= "2021-12-31"),
    "Fuel_Index"
] = 90

fuel.loc[
    (fuel["Date"] >= "2022-02-24") &
    (fuel["Date"] <= "2023-06-30"),
    "Fuel_Index"
] = 145

fuel.loc[
    (fuel["Date"] >= "2023-07-01") &
    (fuel["Date"] <= "2024-12-31"),
    "Fuel_Index"
] = 125

fuel.loc[
    (fuel["Date"] >= "2025-01-01"),
    "Fuel_Index"
] = 115
```

```python
np.random.seed(42)

fuel["Fuel_Index"] = (
    fuel["Fuel_Index"]
    + np.random.normal(0, 4, len(fuel))
)

fuel["Fuel_Index"] = fuel["Fuel_Index"].round(2)

fuel.to_csv(
    OUTPUT_FOLDER + "fact_fuel_index.csv",
    index=False
)

fuel.head()
```

```python
competition = calendar.copy()

competition["Competition_Index"] = 100
```

```python
competition.loc[
    (competition["Date"] >= "2020-03-01") &
    (competition["Date"] <= "2021-12-31"),
    "Competition_Index"
] = 60

competition.loc[
    (competition["Date"] >= "2022-01-01") &
    (competition["Date"] <= "2023-10-06"),
    "Competition_Index"
] = 90

competition.loc[
    (competition["Date"] >= "2023-10-07") &
    (competition["Date"] <= "2024-12-31"),
    "Competition_Index"
] = 35

competition.loc[
    (competition["Date"] >= "2025-01-01"),
    "Competition_Index"
] = 75
```

```python
competition["Competition_Index"] = (
    competition["Competition_Index"]
    + np.random.normal(0, 3, len(competition))
)

competition["Competition_Index"] = (
    competition["Competition_Index"]
    .clip(lower=0, upper=100)
    .round(2)
)

competition.to_csv(
    OUTPUT_FOLDER + "fact_competition.csv",
    index=False
)

competition.head()
```
