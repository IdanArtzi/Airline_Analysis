```python
import pandas as pd
import numpy as np

OUTPUT_FOLDER = "../Data/Generated/"
```

```python
# Disabling scientific notation globally in Pandas
pd.set_option('display.float_format', '{:,.2f}'.format)
```

```python
fact = pd.read_csv(
    OUTPUT_FOLDER +
    "fact_financials.csv"
)

print(fact.shape)
```

```python
# Validating columns
fact.columns.tolist()
```

```python

fact["Date"] = pd.to_datetime(
    fact["Date"]
)
```

```python
# Execetive KPIs
kpi_summary = pd.DataFrame({

    "Metric":[
        "Revenue",
        "Operating Cost",
        "Operating Profit",
        "Passengers",
        "Flights Operated",
        "Average Load Factor"
    ],

    "Value":[

        fact["Total_Revenue_USD"].sum(),

        fact["Total_Operating_Cost_USD"].sum(),

        fact["Operating_Profit_USD"].sum(),

        fact["Passengers"].sum(),

        fact["Flights_Operated"].sum(),

        fact["Load_Factor"].mean()

    ]
})

kpi_summary
```

```python
# Route profitability
route_profitability = (
    fact
    .groupby("City")
    .agg({

        "Total_Revenue_USD":"sum",
        "Total_Operating_Cost_USD":"sum",
        "Operating_Profit_USD":"sum",
        "Passengers":"sum",
        "Load_Factor":"mean"

    })
    .reset_index()
)

route_profitability = (
    route_profitability
    .sort_values(
        "Operating_Profit_USD",
        ascending=False
    )
)

route_profitability.head(10)
```

```python
# Country performance
country_summary = (
    fact
    .groupby("Country")
    .agg({

        "Total_Revenue_USD":"sum",
        "Operating_Profit_USD":"sum",
        "Passengers":"sum"

    })
    .reset_index()
)

country_summary.sort_values(
    "Total_Revenue_USD",
    ascending=False
).head(10)
```

```python
# Region performance
region_summary = (
    fact
    .groupby("Region")
    .agg({

        "Total_Revenue_USD":"sum",
        "Operating_Profit_USD":"sum",
        "Passengers":"sum"

    })
    .reset_index()
)

region_summary
```

```python
# Monthly trend
fact["Year"] = fact["Date"].dt.year
fact["Month"] = fact["Date"].dt.month

monthly_summary = (
    fact
    .groupby(
        ["Year", "Month"]
    )
    .agg({
        "Total_Revenue_USD": "sum",
        "Operating_Profit_USD": "sum",
        "Passengers": "sum"
    })
    .reset_index()
)

monthly_summary.head()
```

```python
route_profitability.head(10)
```

```python
# Load Factor by route
load_factor_routes = (
    fact
    .groupby("City")
    .agg({

        "Load_Factor":"mean"

    })
    .reset_index()
)

load_factor_routes = (
    load_factor_routes
    .sort_values(
        "Load_Factor",
        ascending=False
    )
)

load_factor_routes.head(10)
```

```python
fact.columns.tolist()
```

```python
# Profitability by Route

route_profit = (
    fact
    .groupby("City")
    .agg({

        "Operating_Profit_USD":"sum",
        "Total_Revenue_USD":"sum"

    })
    .reset_index()
)

route_profit["Margin_%"] = (
    route_profit["Operating_Profit_USD"]
    /
    route_profit["Total_Revenue_USD"]
    * 100
)

route_profit = route_profit.sort_values(
    "Margin_%",
    ascending=False
)

route_profit.head(10)
```

```python
# Exporting Final Analysis Tables
kpi_summary.to_csv(
    OUTPUT_FOLDER +
    "kpi_summary.csv",
    index=False
)

route_profitability.to_csv(
    OUTPUT_FOLDER +
    "route_profitability.csv",
    index=False
)

country_summary.to_csv(
    OUTPUT_FOLDER +
    "country_summary.csv",
    index=False
)

region_summary.to_csv(
    OUTPUT_FOLDER +
    "region_summary.csv",
    index=False
)

monthly_summary.to_csv(
    OUTPUT_FOLDER +
    "monthly_summary.csv",
    index=False
)

load_factor_routes.to_csv(
    OUTPUT_FOLDER +
    "load_factor_routes.csv",
    index=False
)

print("Analysis tables saved")
```
