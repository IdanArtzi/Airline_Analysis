```python
# ==========================================
# Generate dim_aircraft
# ==========================================

aircraft_rows = []

for idx, (aircraft_name, values) in enumerate(AIRCRAFT.items(), start=1):
    
    aircraft_rows.append({
        "Aircraft_ID": idx,
        "Aircraft_Type": aircraft_name,
        "Seats": values["seats"],
        "Fuel_Burn_Index": values["fuel_burn_index"],
        "Daily_Lease_USD": values["daily_lease_usd"]
    })

dim_aircraft = pd.DataFrame(aircraft_rows)

print(dim_aircraft.head())
```

```python
# ==========================================
# Generate dim_route
# ==========================================

route_rows = []

for idx, route in enumerate(ROUTES, start=1):

    route_rows.append({
        "Route_ID": idx,
        "City": route["city"],
        "Country": route["country"],
        "Region": route["region"],
        "Distance_KM": route["distance_km"],
        "Route_Tier": route["tier"],
        "Seasonal_Route": route["seasonal"],
        "Launch_Date": route["launch_date"],
        "Base_Daily_Demand": route["base_demand"],
        "Fare_Low_USD": route["fare_low"],
        "Fare_High_USD": route["fare_high"],
        "Aircraft_Type": route["aircraft"],
        "Base_Daily_Flights": route["base_daily_flights"],
        "Competition_Sensitivity": route["competition_sensitivity"]
    })

dim_route = pd.DataFrame(route_rows)

print(dim_route.head())
```

```python
dim_route.to_csv(
    OUTPUT_FOLDER + "dim_route.csv",
    index=False
)

print("dim_route.csv created")
```

```python
# ==========================================
# Generate dim_date
# ==========================================

dates = pd.date_range(
    start="2020-01-01",
    end="2025-12-31",
    freq="D"
)

dim_date = pd.DataFrame({
    "Date": dates
})

dim_date["Year"] = dim_date["Date"].dt.year
dim_date["Quarter"] = dim_date["Date"].dt.quarter
dim_date["Month"] = dim_date["Date"].dt.month
dim_date["Month_Name"] = dim_date["Date"].dt.month_name()
dim_date["Week_Number"] = dim_date["Date"].dt.isocalendar().week
dim_date["Day_Of_Week"] = dim_date["Date"].dt.day_name()

dim_date["Is_Weekend"] = (
    dim_date["Date"].dt.dayofweek >= 5
)
```

```python
# ==========================================
# Season
# ==========================================

def assign_season(month):

    if month in [12, 1, 2]:
        return "Winter"

    elif month in [3, 4, 5]:
        return "Spring"

    elif month in [6, 7, 8]:
        return "Summer"

    else:
        return "Fall"

dim_date["Season"] = dim_date["Month"].apply(assign_season)
```

```python
dim_date.to_csv(
    OUTPUT_FOLDER + "dim_date.csv",
    index=False
)

print("dim_date.csv created")
```
